# Beamforming visualised

This notebook presents visualisations how different kinds of beamformers work. The aim of this notebook is to help build understanding and intuition for what beamformers actually do. This is motivated by several discussions I had over the years that revealed a great deal of confusion in the seismological community. 

Beamformers can be classified along different axes. For the considerations here, two axes are useful to keep in mind:

1. **Propagation model**: How do you expect the wavefield to propagate across your array? Alternatively, what kind of propagation is resolvable? In Seismology, we usually assume plane-waves propagating aross the array as we investigate sources far away. This is also what seismologists commonly mean when they just say "beamforming", which drives large part of the confusion. When sources are close and plane-waves are inadequate descriptions, curved wavefronts can be easily employed. For this approach, the term "Matched Field Processing" has become the standard. For ease of communication, I would prefer "curved-wave beamforming".

2. **Phase-match evaluation**: Given the propagation model, how does the beamformer evaluate what is a good propagation model and what is not? This is the actual beamformer itself. Any kind of propagation model can be combined with any kind of beamformer. Beamformers are commonly classified into

The beamformers presented here are:

### Delay-and-sum

The most straightforward and intuitive beamformer. It corresponds to phase-shifting the waveforms $d(\omega,j)$ recorded on all sensors $j$ for candidate waves. The beampower $B$ is given by

$B = \sum_\omega \sum_{j} | d(\omega,j) s(\omega,j) |$

with $d(\omega,j)$ the data (frequency-domain) recorded at sensor $j$

and

$s(\omega,j) = e^{-i \omega t}$ the steering vector (Greens' function, synthetic wavefield, ...)

### Bartlett beamforming 

Delay-and-sum applied to inter-sensor correlation functions.

$B = \sum_\omega \sum_{j,k} | S(\omega,j,k) C(\omega,j,k) |$

with 

$C(\omega,j,k) = d^*(j,\omega) d(k, \omega)$ 

and 

$S(\omega,j,k) = s(j,\omega) s^*(k, \omega)$

### Cross-correlation beamforming 

Delay-and-sum applied to inter-sensor correlation functions excl. autocorrelations.

$B = \sum_\omega \sum_{j \neq k} S(\omega,j,k) C(\omega,j,k)$

### Minimum Variance Distortionless Response (MVDR) 

$B = \sum_\omega \sum_{j, k} \frac{1}{S(\omega,j,k) C^{-1}(\omega,j,k)}$

### Multiple Signal Classification (MUSIC)

Sub-space beamformer that steers the noise subspace

$B = \sum_\omega \sum_{j,k} | S(\omega,j,k) C_N(\omega,j,k) |$
